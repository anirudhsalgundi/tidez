```python
def create_lightcurve(fits_file):
    with fits.open(fits_file) as hdul:
        data = hdul[1].data

    # 1. separate scalar columns from multi-dim image-stamp columns
    scalar_cols = [name for name in data.columns.names if data[name].ndim == 1]
    stamp_cols  = [name for name in data.columns.names if data[name].ndim > 1]

    # 2. build a Table from the FULL data (astropy Table can hold the stamp columns)
    full_table = Table(data)

    # 3. select just the scalar columns for your "flat" table
    table = full_table[scalar_cols]

    mjd = table["MJD_OBS"]
    mag = table['MAG_FPHOT']
    err = table['MAGERR_FPHOT']
    lim3 = table['LIM_MAG3']
    lim5 = table['LIM_MAG5']
    snr = table['SNR_FPHOT']
    cnn = table['CNNSCORE_ALT']
    filt = table['FILTER']
    sub_quality = table['STATUS_FPHOT']

    good_detections_mask = (snr > 5) & (cnn > 0.5) & (sub_quality == "p")
    upper_limits_mask = (mag > lim5) | np.isnan(mag)

    detections = {
        "mjd": mjd[good_detections_mask],
        "mag": mag[good_detections_mask],
        "err": err[good_detections_mask],
        "filt": filt[good_detections_mask],
    }

    upper_limits = {
        "mjd": mjd[upper_limits_mask],
        "lim5": lim5[upper_limits_mask],
        "filt": filt[upper_limits_mask],
    }

    lightcurve = {
        "detections": detections,
        "upper_limits": upper_limits,
    }

    n_dropped = (~good_detections_mask & ~upper_limits_mask).sum()
    print(f"{n_dropped} of {len(mjd)} rows fell into neither detections nor upper_limits")
    return lightcurve
```

```python
def plot_lightcurve(lightcurve: dict):

    colors = {'g': 'green', 'r': 'red', 'i': 'blue', 'z': 'purple', 'Y': 'orange'}


    detections = lightcurve["detections"]
    upper_limits = lightcurve["upper_limits"]

    plt.figure(figsize=(10, 6))
    # Plot detections
    for filt in np.unique(detections["filt"]):
        mask = detections["filt"] == filt
        plt.errorbar(detections["mjd"][mask], detections["mag"][mask], yerr=detections["err"][mask],
                     fmt='o', label=f'Detections ({filt})', color=colors.get(filt, 'black'))

    for filt in np.unique(upper_limits["filt"]):
        mask = upper_limits["filt"] == filt
        plt.scatter(upper_limits["mjd"][mask], upper_limits["lim5"][mask],
                    marker='v', label=f'Upper Limits ({filt})', alpha=0.5, color=colors.get(filt, 'black'))

    plt.gca().invert_yaxis()  # Magnitude scale is inverted
    plt.xlabel('MJD')
    plt.ylabel('Magnitude')
    plt.title('Lightcurve')
    plt.legend()
    plt.show()
    
```

```python
import requests
from io import BytesIO
from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# RA, DEC = 73.3802607540275, -22.783294013889787
RA, DEC = 73.3802607540275, -22.783294013889787

# Legacy Survey cutout service
size_pix = 150          # image size in pixels
pixscale = 0.262        # arcsec/pixel (native DECam scale)
layer = "ls-dr10"        # legacy survey data release layer

cutout_url = (
    f"https://www.legacysurvey.org/viewer/cutout.jpg"
    f"?ra={RA}&dec={DEC}&layer={layer}"
    f"&pixscale={pixscale}&size={size_pix}"
)

resp = requests.get(cutout_url)
resp.raise_for_status()

img = Image.open(BytesIO(resp.content))

fig, ax = plt.subplots(figsize=(6, 6))
ax.imshow(img)

# center of image = target RA/Dec
cx, cy = size_pix / 2, size_pix / 2

# crosshair
crosshair_len = 20
ax.plot([cx - crosshair_len, cx - 5], [cy, cy], color="red", lw=1.5)
ax.plot([cx + 5, cx + crosshair_len], [cy, cy], color="red", lw=1.5)
ax.plot([cx, cx], [cy - crosshair_len, cy - 5], color="red", lw=1.5)
ax.plot([cx, cx], [cy + 5, cy + crosshair_len], color="red", lw=1.5)

# circle marker (radius in pixels, e.g. corresponding to ~3 arcsec)
circle_radius_pix = 3 / pixscale
circ = patches.Circle((cx, cy), circle_radius_pix, fill=False, color="lime", lw=1.5)
ax.add_patch(circ)

ax.set_title(f"RA={RA:.5f}, Dec={DEC:.5f}")
ax.axis("off")
plt.show()
```

```python
import requests
from astropy.table import Table
from astropy.io import fits
from io import BytesIO

RA, DEC = 73.3802607540275, -22.783294013889787

# tiny box just around the exact position (~2 arcsec half-width)
box_deg = 2 / 3600.0

cat_url = (
    f"https://www.legacysurvey.org/viewer/ls-dr10/cat.fits"
    f"?ralo={RA - box_deg:.6f}&rahi={RA + box_deg:.6f}"
    f"&declo={DEC - box_deg:.6f}&dechi={DEC + box_deg:.6f}"
)

resp = requests.get(cat_url)
resp.raise_for_status()

hdul = fits.open(BytesIO(resp.content))
tab = Table(hdul[1].data)

if len(tab) == 0:
    print("No Legacy Survey source found at this position.")
else:
    # closest match to the exact RA/Dec
    import numpy as np
    dist = np.hypot(tab["ra"] - RA, (tab["dec"] - DEC))
    row = tab[np.argmin(dist)]
    print(row)
```

```python
from astroquery.ipac.irsa import Irsa
from astropy.coordinates import SkyCoord
import astropy.units as u

RA, DEC = 73.3802607540275, -22.783294013889787

coord = SkyCoord(ra=RA, dec=DEC, unit="deg", frame="icrs")

# AllWISE Source Catalog cone search
radius = 5 * u.arcsec

result = Irsa.query_region(
    coordinates=coord,
    catalog="allwise_p3as_psd",   # AllWISE Source Catalog
    spatial="Cone",
    radius=radius,
)

if len(result) == 0:
    print("No WISE match found within radius.")
else:
    cols = ["designation", "ra", "dec", "w1mpro", "w2mpro", "w3mpro", "w4mpro"]
    cols = [c for c in cols if c in result.colnames]
    result[cols].pprint()
```

```python
import requests
import matplotlib.pyplot as plt
from astropy.io import fits
from astropy.visualization import astropy_mpl_style

# 1. Define target parameters (Example: NGC 4535)

SIZE_PX = 500
PIXSCALE = 0.262
LAYER = "ls-dr10"
BANDS = "g"

# 2. Construct the Legacy Survey DR10 FITS API URL
base_url = "https://legacysurvey.org"
params = {
    "ra": RA,
    "dec": DEC,
    "size": SIZE_PX,
    "pixscale": PIXSCALE,
    "layer": LAYER,
    "bands": BANDS
}

print(f"Requesting DR10 FITS cutout for RA={RA}, DEC={DEC}...")

# 3. Download the FITS file
response = requests.get(base_url, params=params)

if response.status_code == 200:
    filename = f"legacy_survey_dr10_{BANDS}.fits"
    with open(filename, "wb") as f:
        f.write(response.content)
    print(f"Successfully saved to {filename}")
    
    # 4. Open and display the FITS image using astropy
    plt.style.use(astropy_mpl_style)
    with fits.open(filename) as hdul:
        # Print file structure information
        hdul.info()
        
        # Get image data
        img_data = hdul[0].data
        
        # Plot the data
        plt.figure(figsize=(6, 6))
        plt.imshow(img_data, cmap="viridis", origin="lower")
        plt.colorbar(label="Flux")
        plt.title(f"DR10 Cutout ({BANDS}-band) at RA={RA}, DEC={DEC}")
        plt.grid(False)
        plt.show()
else:
    print(f"Failed to fetch image. HTTP Status Code: {response.status_code}")
```
